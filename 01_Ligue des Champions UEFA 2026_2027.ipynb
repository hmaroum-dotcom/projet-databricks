**DF_matches**

```python
# 1. Importation des fonctions de Databricks / PySpark
from pyspark.sql import functions as F
from pyspark.sql.types import StructType, StructField, StringType, IntegerType

# 1. Définition de la structure des données
schema = StructType([
    StructField("match_id", IntegerType(), True),
    StructField("home_team", StringType(), True),
    StructField("away_team", StringType(), True),
    StructField("home_goals", IntegerType(), True),
    StructField("away_goals", IntegerType(), True)
])

# 2. Simulation de 24 matchs impliquant de nombreuses équipes européennes
donnees_champions_league = [
    (1, "Paris SG", "Real Madrid", 3, 1),
    (2, "Bayern Munich", "Arsenal", 2, 2),
    (3, "FC Barcelone", "Monaco", 1, 2),
    (4, "Manchester City", "Juventus", 4, 0),
    (5, "Inter Milan", "Liverpool", 0, 2),
    (6, "Bayer Leverkusen", "AC Milan", 2, 1),
    (7, "Atletico Madrid", "RB Leipzig", 1, 0),
    (8, "Borussia Dortmund", "Sporting CP", 3, 2),
    (9, "Benfica", "Feyenoord", 1, 1),
    (10, "PSV Eindhoven", "Girona", 2, 0),
    (11, "Aston Villa", "Bologna", 2, 1),
    (12, "Celtic Glasgow", "Club Brugge", 0, 0),
    (13, "Stuttgart", "Lille OSC", 1, 2),
    (14, "Eintracht Frankfurt", "Lazio", 2, 3),
    (15, "Real Sociedad", "AS Roma", 1, 1),
    (16, "Shakhtar Donetsk", "Dinamo Zagreb", 0, 2),
    (17, "Salzburg", "Brest", 1, 3),
    (18, "Sparta Prague", "Young Boys", 2, 0),
    (19, "Sturm Graz", "Sporting CP", 0, 2),
    (20, "Crvena Zvezda", "Monaco", 1, 4),
    (21, "Real Madrid", "Borussia Dortmund", 5, 2),
    (22, "Paris SG", "PSV Eindhoven", 1, 1),
    (23, "Juventus", "VfB Stuttgart", 0, 1),
    (24, "Arsenal", "Shakhtar Donetsk", 1, 0)
]

# 3. Création du DataFrame global
df_matches = spark.createDataFrame(donnees_champions_league, schema)

# 4. Calcul des points à domicile et à l'extérieur
df_clean = df_matches.withColumn(
    "home_points", 
    F.when(F.col("home_goals") > F.col("away_goals"), 3)
     .when(F.col("home_goals") == F.col("away_goals"), 1)
     .otherwise(0)
).withColumn(
    "away_points",
    F.when(F.col("away_goals") > F.col("home_goals"), 3)
     .when(F.col("away_goals") == F.col("home_goals"), 1)
     .otherwise(0)
)

print("Nouveau jeu de données de la Ligue des Champions chargé avec succès !")

```

**DATAFRAMES**

```python
# 1. Extraction des statistiques pour les équipes à domicile
df_home = df_clean.select(
    F.col("home_team").alias("team"),
    F.col("home_points").alias("points"),
    F.col("home_goals").alias("goals_scored"),
    F.col("away_goals").alias("goals_conceded")
)

# 2. Extraction des statistiques pour les équipes à l'extérieur
df_away = df_clean.select(
    F.col("away_team").alias("team"),
    F.col("away_points").alias("points"),
    F.col("away_goals").alias("goals_scored"),
    F.col("home_goals").alias("goals_conceded")
)

# 3. Fusion des deux tableaux (Domicile + Extérieur)
df_all_matches = df_home.union(df_away)

# 4. Calcul du classement agrégé par équipe
df_ranking = df_all_matches.groupBy("team").agg(
    F.count("points").alias("matches_played"),
    F.sum("points").alias("total_points"),
    F.sum("goals_scored").alias("total_scored"),
    F.sum("goals_conceded").alias("total_conceded")
)

# 5. Ajout de la différence de buts et tri selon les règles de l'UEFA
df_final_ladder = df_ranking.withColumn(
    "goal_difference", 
    F.col("total_scored") - F.col("total_conceded")
).orderBy(
    F.col("total_points").desc(),      # Équipe avec le plus de points en premier
    F.col("goal_difference").desc(),   # En cas d'égalité : meilleure différence de buts
    F.col("total_scored").desc()       # En cas d'égalité secondaire : meilleure attaque
)

# 6. Affichage du classement final
display(df_final_ladder)
```

**Identifier les Qualifiés, Barragistes et Éliminés**

```python
from pyspark.sql.window import Window
from pyspark.sql import functions as F

# 1. Set up the window specification without performance warnings
windowSpec = Window.partitionBy(F.lit(1)).orderBy(
    F.col("total_points").desc(), 
    F.col("goal_difference").desc(), 
    F.col("total_scored").desc()
)

# 2. Generate official rank (1 to 36)
df_with_rank = df_final_ladder.withColumn("Rank", F.row_number().over(windowSpec))

# 3. Assign official UEFA status based on rank
df_uefa_status = df_with_rank.withColumn(
    "UEFA_Status",
    F.when(F.col("Rank") <= 8, "🟢 Qualified (Top 8)")
     .when(F.col("Rank") <= 24, "🟡 Knockout Play-offs (9-24)")
     .otherwise("🔴 Eliminated (25-36)")
)

# 4. Final selection for clean visualization
df_final_display = df_uefa_status.select(
    F.col("Rank"),
    F.col("team").alias("Team"),
    F.col("total_points").alias("Points"),
    F.col("goal_difference").alias("Goal_Diff"),
    F.col("total_scored").alias("Goals_Scored"),
    F.col("UEFA_Status")
)

display(df_final_display)

```

**Calcul taux efficacite**

```python
# 1. Calculate offensive efficiency (Goals scored / Matches played)
df_efficiency = df_final_ladder.withColumn(
    "Offensive_Efficiency",
    F.round(F.col("total_scored") / F.col("matches_played"), 2)
)

# 2. Sort by highest efficiency
df_efficiency_sorted = df_efficiency.orderBy(F.col("Offensive_Efficiency").desc())

# 3. Select and rename columns cleanly for the output
df_efficiency_display = df_efficiency_sorted.select(
    F.col("team").alias("Team"),
    F.col("matches_played").alias("Matches_Played"),
    F.col("total_scored").alias("Goals_Scored"),
    F.col("Offensive_Efficiency")
)

# 4. Final display in Databricks
display(df_efficiency_display)

```

**PySpark pour l'analyse défensive**

```python
# 1. IMPORTATION DES FONCTIONS SPARK
from pyspark.sql import functions as F
from pyspark.sql.types import StructType, StructField, StringType, IntegerType

# 2. CHARGEMENT DES DONNÉES SOURCE (Indispensable pour l'indépendance du notebook)
schema = StructType([
    StructField("match_id", IntegerType(), True),
    StructField("home_team", StringType(), True),
    StructField("away_team", StringType(), True),
    StructField("home_goals", IntegerType(), True),
    StructField("away_goals", IntegerType(), True)
])

donnees_champions_league = [
    (1, "Paris SG", "Real Madrid", 3, 1), (2, "Bayern Munich", "Arsenal", 2, 2),
    (3, "FC Barcelone", "Monaco", 1, 2), (4, "Manchester City", "Juventus", 4, 0),
    (5, "Inter Milan", "Liverpool", 0, 2), (6, "Bayer Leverkusen", "AC Milan", 2, 1),
    (7, "Atletico Madrid", "RB Leipzig", 1, 0), (8, "Borussia Dortmund", "Sporting CP", 3, 2),
    (9, "Benfica", "Feyenoord", 1, 1), (10, "PSV Eindhoven", "Girona", 2, 0),
    (11, "Aston Villa", "Bologna", 2, 1), (12, "Celtic Glasgow", "Club Brugge", 0, 0),
    (13, "Stuttgart", "Lille OSC", 1, 2), (14, "Eintracht Frankfurt", "Lazio", 2, 3),
    (15, "Real Sociedad", "AS Roma", 1, 1), (16, "Shakhtar Donetsk", "Dinamo Zagreb", 0, 2),
    (17, "Salzburg", "Brest", 1, 3), (18, "Sparta Prague", "Young Boys", 2, 0),
    (19, "Sturm Graz", "Sporting CP", 0, 2), (20, "Crvena Zvezda", "Monaco", 1, 4),
    (21, "Real Madrid", "Borussia Dortmund", 5, 2), (22, "Paris SG", "PSV Eindhoven", 1, 1),
    (23, "Juventus", "VfB Stuttgart", 0, 1), (24, "Arsenal", "Shakhtar Donetsk", 1, 0)
]

df_matches = spark.createDataFrame(donnees_champions_league, schema)

# Re-génération du classement de base (total_conceded, matches_played)
df_home = df_matches.select(F.col("home_team").alias("team"), F.col("home_goals").alias("goals_scored"), F.col("away_goals").alias("goals_conceded"))
df_away = df_matches.select(F.col("away_team").alias("team"), F.col("away_goals").alias("goals_scored"), F.col("home_goals").alias("goals_conceded"))
df_final_ladder = df_home.union(df_away).groupBy("team").agg(
    F.count("goals_scored").alias("matches_played"),
    F.sum("goals_conceded").alias("total_conceded")
)

# 3. CALCUL DE L'ANALYSE DÉFENSIVE (Votre code d'origine)
df_defense = df_final_ladder.withColumn(
    "Defensive_Ratio",
    F.round(F.col("total_conceded") / F.col("matches_played"), 2)
)

# 4. TRI DE LA MEILLEURE À LA PLUS FRAGILE DÉFENSE (Ordre croissant)
df_defense_sorted = df_defense.orderBy(F.col("Defensive_Ratio").asc())

# 5. SÉLECTION ET RENOMMAGE DES COLONNES POUR L'AFFICHAGE EN ANGLAIS
df_defense_display = df_defense_sorted.select(
    F.col("team").alias("Team"),
    F.col("matches_played").alias("Matches_Played"),
    F.col("total_conceded").alias("Goals_Conceded"),
    F.col("Defensive_Ratio").alias("Goals_Conceded_Per_Match")
)

display(df_defense_display)

```
